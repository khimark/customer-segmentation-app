## Understanding Values

A `Value` is a piece of information that you are working with

Every value in Python has a data type

A `Data type` gets to determine the kind of information represented and also the operation that can be performed on it 

### Types of Data Types

1. Integer (int) - whole numbers - 25
2. Float (float) decimal numbers -67.5
3. String (str) - text - "Melody"
4. Boolean (bool) - True or False - True

NB: Quotation marks indicate that a value is a string, even when the characters inside them look like numbers 

## Understanding Variables

A `variable` is a container that stores a value or
a named reference to a value store in a program



```python
name ="Mark"
age = 30
score = 45.5
has_passed = True
```

```python
print(name, age,score,has_passed)
```

## Updating Variables

A variable can be a ssigned a new value

```python
age = 20
print(age)
```

## Inspect Data Types

Python provides the `type()` function to identify the data of a value

```python
print(type(name))
print(type(age))
print(type(score))
print(type(has_passed))
```

## Variable Naming Rules

1. Names must start with a letter or an underscore
2. Names can't start with a number
3. Spaces aren't allowed
4. Names can contain letters, numbers, underscores
5. Variable names are case-sensitive
6. Python reserved keywords can't be used as variable names

Python uses `snake_case` for variable names

```python
student_name = "Asma"
score = 85.0
student_age = 24
course = "Data Science"
```

## Type Casting
`Type casting` is the process of converting a value from one data type to another

```python
age = "25"

age = int(age)

print(age + 5)
```

### Common Type Conversion Functions

`int()` - convert to an integer
`float()` - convert to a float
`str()` - convert to a string
`bool()` - conert to a boolean
