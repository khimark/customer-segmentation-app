### Fruitful and Void Functions

#### Fruitful Functions

Return a value using `return`

Results can be stored in variables

Can be used in other calculations

````python
def calculate_discount(price, discount):
    discount_amount = price * (discount/100)
    final_price = price - discount_amount
    
    return final_price

result = calculate_discount(1500, 15)
print("Final price:", result)
````

````python
# or
def calculate_discount(price, discount_rate):
    discount = price * discount_rate / 100
    return discount


def calculate_final_price(price, discount_rate):
    """Return the price after applying the discount."""
    discount = calculate_discount(price, discount_rate)
    final_price = price - discount
    return final_price


# Get input from the user
price = float(input("Enter the original price: "))
discount_rate = float(input("Enter the discount percentage: "))

discount = calculate_discount(price, discount_rate)
final_price = calculate_final_price(price, discount_rate)

print(f"Discount amount: {discount:.2f}")
print(f"Final price: {final_price:.2f}")
````

## Lambda Functions

A `lambda function` is a small anonymus function defined using the `lambda` keyword

Unlike regular functions, lambda functions are written in a single expression and do not require a function name in their definition

**Syntax**
```python
lambda arguments: expression
```

### Example: Squaring a Number

````python
# Using a regular function
def square(number):
    return number ** 2

result = square(4)
print("Result:", result)
````

````python
# Using a lamda function

square = lambda number: number ** 2

print(square(4))
````

````python
# Using lambda with multiple arguments

add = lambda  a, b: a + b

add(10, 20)
````

### Lambda with `map()`

The `map()` function applies a function to every item in an iterable

````python
numbers = [1,2,3,4,5]

square = list(map(lambda digit: digit ** 2, numbers))

print(square)
````

### Lambda with `filter()` 

````python
numbers = [10,15,20,25,30,35]

even_numbers = list(filter(lambda x: x % 2 == 0, numbers))

print(even_numbers)
````
