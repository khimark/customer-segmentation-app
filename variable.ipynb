 Python provides two mechanisms for controlling program execution

 -`Conditional statements`: allow the program to make decisions
 -`Loops`: allow the programs to repeat instructions

## Conditional Statements
`Conditional statements` allow a program to execute different instructions depending on whether a condition evaluates to `True` or `False`

### The `if` Statement
An `if` statement executes a block of code only when the condition is `True`

**Syntax**
``python
if condition:


**Key concepts**
- A `condition` - is an expression that evaluates to `True` or `False`
- The colon`(:)` - marks the beginning of the conditional block
- `Indentation` - identifying the instruction belonging to the condition
- If the condition is `False`, python skips the instructions inside the block

## The `if-else` Statement
Sometimes you might need a program to choose between two possible outcomes

An `if-else` statement provides two possible execution paths:
- The `if` block executes when the condition is `True`
- The `else` block executes when the condition is `False`

```python
# if-else condition
age = int(input("Enter your age: "))
if age >= 18:
    print("You are eligible to vote.")
else:
    print("You are not eligible to vote.")
```

```python
# Checking whether a number is odd or even
number = int(input("Enter number:"))

if number %2 ==0:
    print(f"{number}, is even.")
else:
    print(f"{number}, is odd")
```

## Chained Conditionals `if-elif-else
Allows you to choose from than two possibilities


```python
# grading
score=float(input("Enter your score."))

if score >=90:
    print("A.")
elif score >=80:
    print("B.")
elif score >=70:
    print("C.")
elif score >=60:
    print("D.")
else:
    print("F.")

    
```

NB: Python evaluates the conditions from top to bottom
The first condition that evalauates to `True` executes, the remaining branches are skipped

## Nested Conditionals

A `Nested Conditional` is an `if` statement placed inside another conditional statement

Imagine this: A student wants to access an examination
conditions; 
1. Be registered
2. Have paid required fee
3. Both requirements must be satisfied

```python
registered = True
fees_paid = False

if registered:
    if fees_paid:
        print("You are registered and your fees are paid")
    else:
        print("You are registered but your fees are not paid")
else:
    print("You are not registered.")
    
    
```

## Short-Ciruit Evaluation and the Guardian Pattern

Python evaluates logical expressions from left to right
With `and`, if the first condition is `False`, pyhon does not evaluate the remaining conditions

This is called `short-circuit evaluation
It can help prevent errors when an operation depends on another condition being satisfied first
