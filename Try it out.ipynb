Write a program that generates a random number between 1 and 100. The user has to guess the number in a limited number of attempts. Provide feedback after each guess (higher, lower, or correct).


```python
import random

secret_number = random.randint(1, 100)

max_attempts = 7
attempts = 0

print("Welcome to the Number Guessing Game!")
print("I have chosen a number between 1 and 100.")
print(f"You have {max_attempts} attempts to guess it.\n")

while attempts < max_attempts:
    try:
        guess = int(input(f"Attempt {attempts + 1}/{max_attempts} - Enter your guess: "))
    except ValueError:
        print("Invalid input. Please enter a whole number.\n")
        continue

    attempts += 1

    if guess == secret_number:
        print(f"Correct! You guessed the number in {attempts} attempt(s).")
        break
    elif guess < secret_number:
        print("Too low! Try a higher number.\n")
    else:
        print("Too high! Try a lower number.\n")

else:
    print(f"Sorry, you have used all {max_attempts} attempts.")
    print(f"The correct number was {secret_number}.")
```

Write a program that allows the user to enter two numbers and an operator (+, -, *, /) and perform the corresponding calculation. Use conditional statements to handle different operators.


```python
# Get the two numbers from the user
number1 = float(input("Enter the first number: "))

number2 = float(input("Enter the second number: "))

# Get the operator
operator = input("Enter an operator (+, -, *, /): ")

# Perform the calculation using conditional statements
if operator == "+":
    result = number1 + number2
    print(f"{number1} + {number2} = {result}")

elif operator == "-":
    result = number1 - number2
    print(f"{number1} - {number2} = {result}")

elif operator == "*":
    result = number1 * number2
    print(f"{number1} * {number2} = {result}")

elif operator == "/":
    if number2 == 0:
        print("Error: Division by zero is not allowed.")
    else:
        result = number1 / number2
        print(f"{number1} / {number2} = {result}")

else:
    print("Invalid operator. Please enter +, -, *, or /.")
```

Write a program that uses a while loop to check if a number entered by the user is a prime number (divisible only by 1 and itself).


```python
# Check whether a number is prime using a while loop

number = int(input("Enter a number: "))

is_prime = True

if number < 2:
    is_prime = False
else:
    divisor = 2

    while divisor * divisor <= number:
        if number % divisor == 0:
            is_prime = False
            break
        divisor += 1

if is_prime:
    print(f"{number} is a prime number.")
else:
    print(f"{number} is not a prime number.")
```

Write a program that uses a for loop to print the multiplication table of a number entered by the user (up to a specified number of multiples).


```python
# Print the multiplication table of a number

number = int(input("Enter a number: "))
multiples = int(input("Enter the number of multiples: "))

print(f"\nMultiplication table for {number}:")

for i in range(1, multiples + 1):
    result = number * i
    print(f"{number} x {i} = {result}")
```

Write functions to calculate the area of different shapes like square (area_of_square(side_length)), rectangle (area_of_rectangle(length, width)), and circle (area_of_circle(radius)).


```python
import math


def area_of_square(side_length):
    """Calculate the area of a square."""
    return side_length ** 2


def area_of_rectangle(length, width):
    """Calculate the area of a rectangle."""
    return length * width


def area_of_circle(radius):
    """Calculate the area of a circle."""
    return math.pi * radius ** 2


# Example usage
print("Square area:", area_of_square(4))
print("Rectangle area:", area_of_rectangle(5, 3))
print("Circle area:", area_of_circle(2))
```

```python
# rounding to 2 decimal
print(f"Circle area: {area_of_circle(2):.2f}")
```

Write a function called calculate_grade that takes a score as input and returns the corresponding letter grade (A, B, C, D, F) based on a grading scale (e.g., A: 90-100, B: 80-89, etc.).

```python
def calculate_grade(score):
    """Return the letter grade for a score from 0 to 100."""
    if 90 <= score <= 100:
        return "A"
    elif 80 <= score <= 89:
        return "B"
    elif 70 <= score <= 79:
        return "C"
    elif 60 <= score <= 69:
        return "D"
    elif 0 <= score <= 59:
        return "F"
    else:
        return None


# Get a score from the user
score = float(input("Enter the score: "))

grade = calculate_grade(score)

if grade is None:
    print("Invalid score. Please enter a number between 0 and 100.")
else:
    print(f"The letter grade is {grade}.")
```
